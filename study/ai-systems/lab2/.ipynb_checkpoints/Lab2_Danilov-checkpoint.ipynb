```python
import warnings
warnings.simplefilter('ignore')

import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import seaborn as sns

sns.set(style='darkgrid')
plt.rcParams['figure.figsize'] = [12, 5]
pd.set_option('display.max_columns', 30, 'display.width', 200)
RANDOM_STATE = 67
N_FOLDS = 5
```

### 1. Данные и отложенная выборка

Повторяю чистку из ЛР1: типы, дубликаты, объявления с 1 июля 2025 (до этого данные собирались неровно), удаление столбцов, где больше 40% пропусков, и строк с пропусками. Затем та же отложенная выборка 80/20 с тем же `random_state`, так что test тот же, что в ЛР1. Вся кросс-валидация дальше идёт только на train, test смотрю один раз для каждого варианта CV — чтобы сравнить, насколько оценка CV похожа на ошибку на новых данных.

Одно отличие от ЛР1: train сортирую по дате публикации. На случайные варианты CV (KFold с перемешиванием, ShuffleSplit) порядок строк не влияет, а варианты для временных рядов (TimeSeriesSplit, KFold без перемешивания) режут выборку по порядку строк, и при сортировке по дате они проверяют модель на будущих объявлениях.

```python
raw = pd.read_csv('../lab1/data/portugal_listinigs.csv', low_memory=False)

df = raw.copy()
df['PublishDate'] = pd.to_datetime(df['PublishDate'], errors='coerce')
for col in ['Garage', 'Elevator', 'ElectricCarsCharging']:
    # в CSV эти столбцы - строки 'True'/'False' вперемешку с bool
    df[col] = df[col].map({True: 1.0, False: 0.0, 'True': 1.0, 'False': 0.0})
df = df.drop_duplicates()
df = df[df['Price'].notna()]
df = df[df['PublishDate'] >= '2025-07-01']
missing = df.isna().mean()
df = df.drop(columns=missing[missing > 0.4].index)
df = df.dropna()
print('после чистки:', df.shape)

from sklearn.model_selection import train_test_split
train_df, test_df = train_test_split(df, test_size=0.2, random_state=RANDOM_STATE)
```

```python
FEATURES_NUM = ['LivingArea', 'NumberOfBathrooms', 'ConstructionYear', 'Parking', 'Elevator']
FEATURES_CAT = ['EnergyCertificate', 'District', 'City']
FEATURES = FEATURES_NUM + FEATURES_CAT
TARGET = 'Price'

def prepare_apartments(d):
    d = d[(d['Type'] == 'Apartment')
          & d['LivingArea'].between(10, 1000)
          & d['Price'].between(10_000, 20_000_000)].copy()
    d.loc[~d['NumberOfBathrooms'].between(0, 10), 'NumberOfBathrooms'] = np.nan
    d.loc[~d['ConstructionYear'].between(1900, 2026), 'ConstructionYear'] = np.nan
    d.loc[d['EnergyCertificate'].isin(['No Certificate', 'Not available']), 'EnergyCertificate'] = 'NC'
    return d

tr = prepare_apartments(train_df).sort_values('PublishDate').reset_index(drop=True)
te = prepare_apartments(test_df).reset_index(drop=True)

X_train, y_train = tr[FEATURES], tr[TARGET]
X_test, y_test = te[FEATURES], te[TARGET]
dates = tr['PublishDate']                          # для оси X на графиках и временных разбиений
districts = tr['District']                         # группы для GroupKFold
months = tr['PublishDate'].dt.strftime('%Y-%m')    # группы для LeaveOneGroupOut

print('train:', X_train.shape, ' test:', X_test.shape)
print('даты train:', dates.min().date(), '—', dates.max().date())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 4))
months.value_counts().sort_index().plot.bar(ax=axes[0], color='tab:blue')
axes[0].set(title='Объявления в train по месяцам', xlabel='', ylabel='объявлений')
districts.value_counts().plot.bar(ax=axes[1], color='tab:green')
axes[1].set(title='Объявления в train по округам', xlabel='', ylabel='объявлений')
plt.tight_layout()
plt.show()
```

По этим двум графикам видно, что группы очень неравные. Объявлений за месяц от 170 летом 2025 до 1600 летом 2026, а на Лиссабон приходится 40% всех квартир, на 8 малых округов вместе — 2%. Это скажется на вариантах CV по группам: фолды по округам получатся разного размера.

### 2. Модель

`LinearRegression` с предобработкой из задания 2.3 ЛР1:
- таргет логарифмирую (`TransformedTargetRegressor`): цена распределена логнормально, на логарифме ошибки становятся относительными, а прогноз возвращается в евро через `exp`;
- площадь тоже логарифмирую: в log-log масштабе цена зависит от площади линейно;
- пропуски числовых признаков заменяю медианой, нормирую `StandardScaler`;
- категории (`EnergyCertificate`, `District`, `City`) кодирую one-hot, редкие значения (меньше 30 объявлений) объединяю, а незнакомые в валидации значения кодируются нулями (`handle_unknown='ignore'`).

Регуляризации нет: по заданию ЛР2 нужна именно линейная регрессия. Вся предобработка внутри `Pipeline`, поэтому на каждом фолде медианы, средние для нормирования и словарь категорий считаются только по train-части фолда — утечки из валидации нет.

```python
from sklearn.base import clone
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, FunctionTransformer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, mean_absolute_error, r2_score

def log_first_column(a):
    a = a.copy()
    a[:, 0] = np.log(a[:, 0])   # 1-й числовой столбец - LivingArea
    return a

num = make_pipeline(SimpleImputer(strategy='median', add_indicator=True),
                    FunctionTransformer(log_first_column, feature_names_out='one-to-one'),
                    StandardScaler())
cat = make_pipeline(SimpleImputer(strategy='most_frequent'),
                    OneHotEncoder(handle_unknown='ignore', min_frequency=30, sparse_output=False))
prep = ColumnTransformer([('num', num, FEATURES_NUM), ('cat', cat, FEATURES_CAT)])

model = TransformedTargetRegressor(regressor=Pipeline([('prep', prep), ('model', LinearRegression())]),
                                   func=np.log, inverse_func=np.exp)

def metrics(y_true, y_pred):
    return {'RMSE': root_mean_squared_error(y_true, y_pred),
            'MAE': mean_absolute_error(y_true, y_pred),
            'R2': r2_score(y_true, y_pred)}

# точка отсчёта: модель на всём train, оценка на test
full = clone(model).fit(X_train, y_train)
FULL_TEST = metrics(y_test, full.predict(X_test))
pd.DataFrame({'train': metrics(y_train, full.predict(X_train)), 'test': FULL_TEST}).T.round({'RMSE': 0, 'MAE': 0, 'R2': 3})
```

### 3. Прогон кросс-валидации и визуализация

`run_cv` делает то же, что функция `cross_validation` из образца: на каждом фолде обучает копию модели на train-части, считает RMSE, MAE и R² на валидации и прогнозирует test. Прогнозы на test усредняются по фолдам, это «TEST» в итоговой таблице, как в образце. Индексы фолдов сохраняются для графика.

`plot_cv` рисует две панели, как в образце:
- слева — каждый объект train как штрих на своей дате публикации, по строке на фолд: синий — train-часть, оранжевый — валидация, серый — объект в этом фолде не используется (такое бывает у ShuffleSplit и вариантов для временных рядов);
- справа — распределение log10(цены) в валидации каждого фолда (оранжевые) и в test (синяя). Если оранжевые кривые похожи на синюю, фолд проверяет модель на данных, похожих на настоящие новые.

Для вариантов по округам по оси X не дата, а объекты, упорядоченные по округу, и нижней строкой показаны сами округа: иначе на графике не видно, что группы целиком уходят в одну часть.

```python
RESULTS = {}

def run_cv(name, cv, groups=None, y_split=None):
    """y_split - метки для стратификации (по умолчанию сам таргет), groups - группы."""
    t0 = time.perf_counter()
    rows, folds, test_pred = [], [], np.zeros(len(X_test))
    for tr_idx, va_idx in cv.split(X_train, y_train if y_split is None else y_split, groups):
        m = clone(model).fit(X_train.iloc[tr_idx], y_train.iloc[tr_idx])
        rows.append(metrics(y_train.iloc[va_idx], m.predict(X_train.iloc[va_idx]))
                    | {'n_train': len(tr_idx), 'n_val': len(va_idx)})
        folds.append((tr_idx, va_idx))
        test_pred += m.predict(X_test)
    fold_df = pd.DataFrame(rows).rename_axis('фолд')
    RESULTS[name] = {'folds': fold_df, 'idx': folds,
                     'test': metrics(y_test, test_pred / len(folds)),
                     'sec': time.perf_counter() - t0}
    s = fold_df
    print(f'{name}: {len(folds)} фолдов, {RESULTS[name]["sec"]:.1f} с')
    print(f'  CV   RMSE = {s.RMSE.mean():,.0f} ± {s.RMSE.std():,.0f} €   R² = {s.R2.mean():.3f} ± {s.R2.std():.3f}')
    print(f'  TEST RMSE = {RESULTS[name]["test"]["RMSE"]:,.0f} €   R² = {RESULTS[name]["test"]["R2"]:.3f}')
    return fold_df.round({'RMSE': 0, 'MAE': 0, 'R2': 3})

def plot_cv(name, by=None):
    """by - группы (например, округ): тогда по оси X объекты упорядочены по группе, а не по дате."""
    folds = RESULTS[name]['idx']
    n = len(folds)
    rows = n + (by is not None)
    fig, ax = plt.subplots(1, 2, gridspec_kw={'width_ratios': [3, 1]}, figsize=(18, 1.2 + 0.45 * rows))
    if by is None:
        x = dates.to_numpy()
    else:
        order = np.lexsort((dates.to_numpy(), by.to_numpy()))
        x = np.empty(len(by))
        x[order] = np.arange(len(by))
    for i, (tr_idx, va_idx) in enumerate(folds):
        unused = np.setdiff1d(np.arange(len(x)), np.concatenate([tr_idx, va_idx]))
        for idx, color, label in [(unused, 'lightgrey', 'не используется'),
                                  (tr_idx, 'tab:blue', 'train'), (va_idx, 'tab:orange', 'валидация')]:
            ax[0].scatter(x[idx], np.full(len(idx), i), marker='|', s=160, lw=0.6, c=color,
                          label=label if i == 0 else None, rasterized=True)
        sns.kdeplot(np.log10(y_train.iloc[va_idx]), color='tab:orange', lw=1, alpha=0.7, ax=ax[1])
    sns.kdeplot(np.log10(y_test), color='tab:blue', lw=2.5, label='test', ax=ax[1])
    ax[0].set(title=name, ylabel='фолд', ylim=(rows - 0.5, -0.5))
    ax[0].set_yticks(range(rows), list(range(n)) + (['группа'] if by is not None else []))
    if by is None:
        ax[0].set_xlabel('дата публикации')
        ax[0].xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
    else:
        # нижняя строка - сами группы, у каждой свой цвет; подписаны группы от 2% объектов
        codes = pd.factorize(by, sort=True)[0]
        ax[0].scatter(x, np.full(len(x), n), marker='|', s=160, lw=0.6, c=codes, cmap='tab20', rasterized=True)
        g = pd.Series(x).groupby(by.to_numpy())
        big = g.size()[g.size() >= 0.02 * len(x)].index
        ax[0].set_xticks(g.mean()[big], big, rotation=30, ha='right', fontsize=9)
        ax[0].set_xlabel(f'объекты train, упорядочены по группе ({by.name})')
    ax[0].legend(loc='upper left', bbox_to_anchor=(1.0, 1.0), fontsize=9, markerscale=0.6)
    ax[1].set(title='log10(цены): валидация фолдов и test', xlabel='log10(цена, €)', ylabel='')
    ax[1].legend(fontsize=9)
    plt.tight_layout()
    plt.show()
```

### 4. Варианты кросс-валидации

Везде 5 фолдов (`N_FOLDS`), где у метода есть такой параметр. Для каждого варианта: что он делает, таблица метрик по фолдам и график фолдов.

#### 1. K-Fold без перемешивания

Train делится на 5 подряд идущих кусков, каждый по очереди становится валидацией, остальные 4 — обучением. Каждый объект ровно один раз попадает в валидацию.

Строки отсортированы по дате, поэтому куски — это периоды времени, и в 4 из 5 фолдов модель учится в том числе на объявлениях из будущего относительно валидации.

```python
from sklearn.model_selection import KFold
cv = KFold(n_splits=N_FOLDS, shuffle=False)
run_cv('K-Fold без перемешивания', cv)
```

```python
plot_cv('K-Fold без перемешивания')
```

Наблюдение.

- R² по фолдам от 0,48 до 0,75, стандартное отклонение 0,108 — втрое больше, чем у K-Fold с перемешиванием (0,033).
- Почти весь разброс даёт фолд 3, объявления с 25.06 по 10.08.2026: RMSE 386 тыс. €, R² 0,48. В этот период попали самые дорогие квартиры train, разбор ниже в разделе 5. Без него четыре фолда дают R² 0,67–0,75.
- Каждый фолд — отдельный период, поэтому оценка показывает, как модель работает в разные месяцы, но модель при этом учится и на будущем.

#### 2. K-Fold с перемешиванием

То же разбиение на 5 частей, но строки сначала перемешиваются, поэтому валидация каждого фолда — случайная пятая часть train, разбросанная по всему периоду. Этот вариант я использовал в ЛР1.

```python
cv = KFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
run_cv('K-Fold с перемешиванием', cv)
```

```python
plot_cv('K-Fold с перемешиванием')
```

Наблюдение.

- R² по фолдам 0,65–0,73, среднее 0,669, CV RMSE 236 тыс. € (± 22 тыс.).
- На графике валидация каждого фолда разбросана по всему периоду, а справа все пять кривых цены почти совпадают с test: фолды похожи друг на друга и на test, поэтому и метрики по фолдам близкие.

#### 3. Repeated K-Fold

K-Fold с перемешиванием, повторённый 3 раза с разными перемешиваниями: 15 фолдов. Каждый объект 3 раза попадает в валидацию. Оценка та же по смыслу, что в варианте 2, но среднее по 15 фолдам меньше зависит от одного удачного или неудачного разбиения.

```python
from sklearn.model_selection import RepeatedKFold
cv = RepeatedKFold(n_splits=N_FOLDS, n_repeats=3, random_state=RANDOM_STATE)
run_cv('Repeated K-Fold (5×3)', cv)
```

```python
plot_cv('Repeated K-Fold (5×3)')
```

Наблюдение.

- Первые 5 фолдов совпадают с K-Fold с перемешиванием: первое повторение использует то же перемешивание при том же `random_state`.
- Среднее R² по трём повторениям 0,669, 0,668 и 0,672: результат одного 5-фолдового прогона отличается от другого всего на ±0,002. При 11 тыс. объектов одного прогона K-Fold достаточно, повторения стоят втрое дороже и почти ничего не дают. Они полезнее на маленьких выборках, где оценка сильно зависит от разбиения.

#### 4. Случайные перестановки (ShuffleSplit)

На каждой итерации заново случайно выбираются 70% объектов для обучения и 20% для валидации, остальные 10% не используются. В отличие от K-Fold, один объект может попасть в валидацию несколько раз или ни разу.

```python
from sklearn.model_selection import ShuffleSplit
cv = ShuffleSplit(n_splits=N_FOLDS, train_size=0.7, test_size=0.2, random_state=RANDOM_STATE)
run_cv('ShuffleSplit (70/20)', cv)
```

```python
plot_cv('ShuffleSplit (70/20)')
```

Наблюдение.

- R² 0,637 ± 0,028, CV RMSE 260 тыс. € — хуже, чем у K-Fold (0,669 и 236 тыс.). Модель здесь учится на 70% train (7846 объектов) вместо 80%, плюс валидационные выборки разных итераций пересекаются, поэтому пять оценок не независимы.
- Серых (неиспользуемых) объектов на графике почти не видно: их 10%, и они перекрыты оранжевыми и синими штрихами соседних объектов.

#### 5. Стратифицированный K-Fold

`StratifiedKFold` придуман для классификации: в каждом фолде та же доля каждого класса, что во всей выборке. Для регрессии делю цену на 10 интервалов по децилям и стратифицирую по номеру интервала. Тогда в каждой валидации одинаковая доля дешёвых и дорогих квартир, и самые дорогие объекты, которые сильнее всего влияют на RMSE, не скапливаются в одном фолде.

```python
from sklearn.model_selection import StratifiedKFold
price_bins = pd.qcut(y_train, 10, labels=False)
cv = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=RANDOM_STATE)
run_cv('Stratified K-Fold (децили цены)', cv, y_split=price_bins)
```

```python
plot_cv('Stratified K-Fold (децили цены)')
```

Наблюдение.

- R² 0,672 ± 0,027 — разброс R² самый маленький среди вариантов с 5 фолдами, но разница с обычным K-Fold небольшая (0,033). Разброс RMSE не уменьшился (± 23 тыс. против ± 22 тыс.).
- Причина: RMSE определяют несколько десятков квартир дороже 3 млн €, а это верхние 0,4% цены. Децили их не разделяют, все они в верхнем интервале, который тоже делится по фолдам случайно. Стратификация по цене работает лучше на маленьких выборках или при более мелком делении хвоста.

#### 6. Group K-Fold по округам

Группа — округ (`District`). Все объявления одного округа попадают или в обучение, или в валидацию. Так проверяется, как модель предсказывает цену в округе, которого она не видела: one-hot столбцы `District` и `City` для таких объявлений нулевые, и модель знает о расположении только то, что заложено в остальных признаках. `GroupKFold` раскладывает 18 округов по 5 фолдам так, чтобы фолды были близки по числу объявлений, но Лиссабон, 40% train, один занимает целый фолд.

```python
from sklearn.model_selection import GroupKFold
cv = GroupKFold(n_splits=N_FOLDS)
run_cv('Group K-Fold (округ)', cv, groups=districts)
```

```python
plot_cv('Group K-Fold (округ)', by=districts)
```

Наблюдение.

- Самая плохая оценка: R² 0,315 ± 0,236, а RMSE по фолдам от 100 до 500 тыс. €.
- Фолд 0 — один Лиссабон: R² −0,035, хуже, чем всем квартирам предсказать среднюю цену. Без Лиссабона в обучении модель не знает надбавку за Лиссабон и Кашкайш и занижает цены.
- Фолд 3 с Фару — RMSE 500 тыс. €: Алгарве целиком умещается в округ Фару, и в обучении этого фолда нет ни одной квартиры из Алгарве, в том числе дорогих.
- Эта оценка отвечает на другой вопрос: как модель работает в округе, которого нет в данных. Для объявлений из известных округов (как в test) она слишком пессимистична. TEST RMSE 224 тыс. € хуже, чем у других вариантов, потому что каждая из пяти моделей не видела часть округов.

#### 7. Group Shuffle Split по округам

Сочетание групп и случайных перестановок: на каждой итерации случайно выбираются 70% округов для обучения и 20% для валидации. Доли считаются по числу округов, а не объявлений, поэтому размер валидации от итерации к итерации сильно меняется — смотря попал в неё Лиссабон или нет.

```python
from sklearn.model_selection import GroupShuffleSplit
cv = GroupShuffleSplit(n_splits=N_FOLDS, train_size=0.7, test_size=0.2, random_state=RANDOM_STATE)
run_cv('Group Shuffle Split (округ)', cv, groups=districts)
```

```python
plot_cv('Group Shuffle Split (округ)', by=districts)
```

Наблюдение.

- CV RMSE 217 тыс. € — меньше, чем у всех остальных вариантов, а R² всего 0,384. Противоречие объясняется тем, какие округа попали в валидацию: Лиссабон и Сетубал не попали ни в одну валидацию (в итерации 0 Лиссабон вообще не используется), а у дешёвых округов маленький RMSE даже при плохом R².
- Размер валидации от 568 до 2844 объектов: доли 70/20 считаются по числу округов.
- Вывод: при неравных группах `GroupShuffleSplit` с 5 итерациями даёт случайную оценку, которая зависит от того, куда попали крупные группы. Средний RMSE по таким фолдам сравнивать с другими нельзя.

#### 8. Leave One Group Out по месяцам

Группа — месяц публикации. Фолдов столько, сколько месяцев (15): валидация — один месяц целиком, обучение — все остальные. Проверяется, держится ли качество модели во времени, или в какие-то месяцы рынок вёл себя иначе.

```python
from sklearn.model_selection import LeaveOneGroupOut
cv = LeaveOneGroupOut()
run_cv('Leave One Group Out (месяц)', cv, groups=months)
```

```python
plot_cv('Leave One Group Out (месяц)')
```

Наблюдение.

- Самый высокий средний R² — 0,691, модель на каждом фолде учится на 14 месяцах из 15 (93% train).
- R² по месяцам от 0,46 до 0,85. Хуже всего июль 2026 (фолд 12): R² 0,46, RMSE 421 тыс. € — тот же выброс, что в K-Fold без перемешивания. Летние месяцы 2025 (фолды 0–2, 170–230 объявлений) тоже дают большой RMSE, 267–412 тыс. €: объявлений мало, и пара дорогих квартир сильно сдвигает метрику.
- С ноября 2025 по июнь 2026 R² 0,63–0,85 без тренда: явного изменения рынка за год модель не видит.

#### 9. Time Series Split

Варианты для временных рядов: валидация всегда позже обучения. `TimeSeriesSplit` делит отсортированный train на 6 кусков; на k-м фолде модель учится на первых k кусках и проверяется на следующем. Обучающая часть с каждым фолдом растёт, а самые поздние объявления в первых фолдах не используются.

Это ближе всего к реальному применению: модель обучают на объявлениях прошлых месяцев и применяют к новым.

```python
from sklearn.model_selection import TimeSeriesSplit
cv = TimeSeriesSplit(n_splits=N_FOLDS)
run_cv('Time Series Split', cv)
```

```python
plot_cv('Time Series Split')
```

Наблюдение.

- R² 0,648 ± 0,111. На фолде 0 модель учится всего на 1869 объявлениях (до 23.01.2026), но R² на следующих 1868 объявлениях 0,662 — почти как с 9341 объявлением на фолде 4 (0,660). Линейной модели с 92 признаками после one-hot хватает пары тысяч объектов.
- Фолд 3 (10.07–18.08.2026) снова худший, R² 0,459: выброс лета 2026.
- Деградации со временем нет: модель, обученная на прошлом, предсказывает следующие недели так же, как случайный K-Fold. Значит, для этих данных случайный K-Fold не завышает качество из-за «подглядывания в будущее».

#### 10. Blocking Time Series Split

Класс из образца: train делится на 5 непересекающихся блоков по времени, в каждом блоке первые 90% — обучение, последние 10% — валидация. Каждый объект используется только в одном фолде, модель учится на одном периоде и проверяется на сразу следующих за ним объявлениях, без данных из других периодов.

```python
class BlockingTimeSeriesSplit:
    def __init__(self, n_splits, train_share=0.9):
        self.n_splits = n_splits
        self.train_share = train_share

    def get_n_splits(self, X=None, y=None, groups=None):
        return self.n_splits

    def split(self, X, y=None, groups=None):
        indices = np.arange(len(X))
        block = len(X) // self.n_splits
        for i in range(self.n_splits):
            start, stop = i * block, (i + 1) * block
            mid = start + int(self.train_share * block)
            yield indices[start:mid], indices[mid:stop]

cv = BlockingTimeSeriesSplit(n_splits=N_FOLDS)
run_cv('Blocking Time Series Split', cv)
```

```python
plot_cv('Blocking Time Series Split')
```

Наблюдение.

- Валидация каждого блока — всего 225 объявлений за несколько дней (по графику это узкие оранжевые полоски), поэтому оценки шумные: R² 0,50–0,70.
- Худший — блок 3 (валидация 5–10.08.2026), RMSE 447 тыс. €: в неё попали две квартиры в Кашкайше по 5,2 и 5,7 млн €, которые модель этого блока оценивает в 1,85 млн.
- Каждая модель учится на 2016 объектах, в 5 раз меньше, чем в K-Fold, а R² почти такой же (0,637). Это ещё одно подтверждение, что модели хватает небольшой выборки.

### Выводы

1. На данных ЛР1 (11 209 квартир в train, `LinearRegression` с логарифмом цены и площади) применил 10 вариантов кросс-валидации: K-Fold без перемешивания и с ним, Repeated K-Fold, ShuffleSplit, Stratified K-Fold по децилям цены, Group K-Fold и Group Shuffle Split по округам, Leave One Group Out по месяцам, Time Series Split и Blocking Time Series Split. Для каждого построена визуализация фолдов (train / валидация / неиспользуемые объекты) и распределения цены в валидации против test.

2. **Случайные разбиения** (K-Fold с перемешиванием, Repeated K-Fold, Stratified K-Fold) дают одинаковую оценку: R² 0,669–0,672, RMSE 236 тыс. €, разброс R² по фолдам около 0,03. Повторения и стратификация на выборке в 11 тыс. объектов почти ничего не меняют: средние трёх повторений отличаются на ±0,002. ShuffleSplit даёт оценку хуже (R² 0,637), потому что модель учится на 70% train, а не на 80%.

3. **Разбиения по времени** (K-Fold без перемешивания, Time Series Split, Blocking Time Series Split, Leave One Group Out по месяцам) в среднем дают ту же оценку, R² 0,637–0,691. Модель, обученная только на прошлом, предсказывает будущие объявления не хуже, чем при случайном разбиении, так что для этих данных перемешивание не завышает качество. Но разброс по фолдам втрое больше (std R² 0,08–0,11): все дорогие квартиры одного пакета попадают в один фолд.

4. **Разбиения по округам** дают самую плохую оценку: R² 0,315 (Group K-Fold) и 0,384 (Group Shuffle Split). Модель не переносится на округ, которого не было в обучении (на Лиссабоне R² −0,035): цену в первую очередь определяет расположение, а его модель берёт из one-hot столбцов округа и города. Group Shuffle Split при таких неравных группах вообще ненадёжен: Лиссабон и Сетубал ни разу не попали в валидацию, поэтому его RMSE (217 тыс. €, самый низкий) с остальными сравнивать нельзя.

5. **Что выбрать для этой задачи.** Модель применяется к новым объявлениям из тех же округов, поэтому основная оценка — K-Fold с перемешиванием или стратифицированный K-Fold на 5 фолдов: быстрые (0,4 с), стабильные и согласованные с test. Time Series Split полезен как дополнительная проверка, что качество не падает со временем. Group K-Fold по округам нужен только если модель будут применять в новых регионах. Leave-One-Out я не делал: это 11 тыс. обучений, а при такой выборке он не даёт ничего сверх 5-фолдовой оценки.

6. **Про метрику.** RMSE сильно зависит от нескольких очень дорогих квартир: 20 объектов меняют RMSE фолда с 250 до 424 тыс. €. Для сравнения вариантов CV надёжнее R² и MAE, а саму модель стоит дополнить удалением выбросов, как в задании 2.3 ЛР1.

7. Все варианты оценивают ошибку выше, чем она оказалась на test (CV R² 0,64–0,69 против 0,715 у случайных и временных вариантов). Причина — состав выборок, а не переобучение: на самом train модель ошибается так же, как на CV.
